```python
def pre_process(df):
    target_cols = 'Survived'

    NUM = ['PassengerId', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
    CAT = ['Sex', 'Embarked']

    df = df[NUM + CAT + [target_cols]]
    df = pd.get_dummies(df, columns=CAT, drop_first=True)
    train , valid = train_test_split(df, test_size=0.2, random_state=42)
    x_train = train.drop(columns=[target_cols])
    y_train = train[target_cols]

    x_valid = valid.drop(columns=[target_cols])
    y_valid = valid[target_cols]

    return x_train, y_train, x_valid, y_valid
```

```python
x_train, y_train, x_valid, y_valid = pre_process(df)
```

```python
lr = LogisticRegression(max_iter=1000)
model = lr.fit(x_train, y_train)
```

```python
def get_metrics(model, x_valid, y_valid):
    preds= model.predict(x_valid)
    return f''' 

Confusion Matrix: {confusion_matrix(y_valid, preds)}

Precision: {precision_score(y_valid, preds, average='weighted', zero_division=0)}

Recall: {recall_score(y_valid, preds, average='weighted', zero_division=0)}

F1 Score: {f1_score(y_valid, preds, average='weighted', zero_division=0)}

Classification Report: {classification_report(y_valid, preds, zero_division=0)}
    '''
```

```python
print(get_metrics(model, x_valid, y_valid))
```
