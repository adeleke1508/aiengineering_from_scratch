```python
def split_df(df):
    target_cols = '2urvived'
    train, test = train_test_split(df)
    x_train = train.drop(columns = [target_cols])
    y_train = train[target_cols]
    x_valid = test.drop(columns = [target_cols])
    y_valid = test[target_cols]
    x_train = x_train.to_numpy()
    y_train = y_train.to_numpy()
    x_valid = x_valid.to_numpy()
    y_valid = y_valid.to_numpy()
    return x_train, y_train, x_valid, y_valid
```

```python
x_train, y_train, x_valid, y_valid = split_df(df)
```

```python
model = LinearSVM()
svm = model.fit(x_train, y_train)
```

```python
def get_metrics(model , y_valid):
    preds = model.predict(x_valid)
    precision = precision_score(y_valid, preds, average='weighted', zero_division=0)
    recall = recall_score(y_valid, preds, average='weighted', zero_division=0)
    f1 = f1_score(y_valid, preds, average='weighted', zero_division=0)
    print(f''' 
Precision : {precision}
Recall: {recall}
F1_score : {f1}     
''')
```

```python
get_metrics(svm, y_valid)
```
